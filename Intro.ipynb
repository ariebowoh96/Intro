```python
job_title = 'Data Analyst'
job_location = 'Miami'
job_salary = 90000
def display_info(title,location,salary):
  return print(f' JOB       : {job_title}\n LOCATION  : {job_location}\n SALARY    : ${job_salary:,.0f}')
```

```python
class job_post:
  def __init__(self,title,location,salary):
    self.title = title
    self.location = location
    self.salary = salary
  def display_info(self):
    return print(f' JOB       : {job_title}\n LOCATION  : {job_location}\n SALARY    : ${job_salary:,.0f}')
```

```python
job_1 = job_post(job_title,job_location,job_salary)
```
